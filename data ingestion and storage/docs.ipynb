## data type
- structured data
    * data that is organized in a defined maner or schema, typicalle found in relational database
    * easily queryable
    * organized in rows and colums
    * Has a consistent structure
    * example: mysql, postgreSQL, csv, excel,...
- unstructured data
    * data that doesn't have predefinded structure or schema
    * not easily queryable without preprocessing
    * various format
    * example: text file, video and audio, image, email,...
- semi-structured data
    * data that is not as organized as struture data but has some lever of structure in the forme format of tag, hierarchies
    * example: xml, json, logfile

## datawarehouse & data lake & data lakehouse
1. data warehouse (schema on write)
    - definition: a centralizaed repository optimize for analysis where data from different source is stored in a structure format
    - design for complex queries and analysis
    - data is cleaned, tranfromed and load
    * example: Redshift, Big query, SQL data warehouse
2. data lake (schema on read)
    - definition: a storage  repository the holds vast amounts of raw data in its native format
    - can store large columes of raw data without predifined schena
    - data is loaded as-is, no need for preprocessing
    - support batch, real-time, and stream processing
    - can be queryed foe data tranformation or exploration purposes
    * example: S3, HDFS
3. data lakehouse
    - Definition: A hybrid data architecture that combines the best features of data lakes and data warehouses, aiming to provide the performance,reliability, and capabilities of a data warehouse while maintaining the flexibility, scale, and low-cost storage of data lakes.

## ETL pipline & orchestrtion
- datawarehouse use ETL
- data lakehouse use ELT
    * E: extract - retrieve raw data
                 - ensure data intergrity during the extract phase
                 - can done is real-time or in batchs, depending on requirements
    * T: tranform - convert the extract data into a format suitable for the target data warehouse
                  -  data cleansing
                  - data enrichment
                  - format changes
                  - aggregation or computatuions
                  - Encoding/ decoding data
                  - handling missing values
    * L: load - move the tranformed data into the target warehouse or another repository
              - bacth/ streaming
              - ensure data intergrity during the load phase

    * this process must be automated in some reliable way

## S3 Object Encryption
- Know the four encryption methods: SSE-S3, SSE-KMS, SSE-C, Client-Side.
- Headers required for SSE-S3 and SSE-KMS.
- KMS quotas can be a bottleneck for high-throughput workloads.
- SSE-C requires HTTPS and customer-provided keys.
- Client-side encryption = full control, AWS only stores encrypted data.
- Encryption in transit enforced via bucket policy with SecureTransport.

    - Default = SSE-S3, but configurable (e.g., SSE-KMS).
    - Bucket policies enforce encryption by rejecting non-compliant requests.
    - Policies take precedence over default encryption.

## S3 access Points
- Access points simplify security management for large buckets.
- Each access point has its own policy + DNS name.
- Can be internet-facing or VPC-only.
- VPC endpoint required for private access, with its own policy.
- Security is layered: bucket policy + access point policy + VPC endpoint policy.

## EBS Volume
- EBS = persistent, network-attached block storage.
- Bound to AZ, but snapshots allow cross-AZ movement.
- One volume → one instance (at CCP level), but an instance can have multiple volumes.
- Delete on termination is critical to understand for exam questions.
    * Elastic Volumes = modify size, type, performance live.
    * Increase only for size (no shrinking).
    * Change type (e.g., GP2 → GP3) and adjust IOPS/throughput on the fly.
    * No downtime or detachment required.

## EFS Elsetic File System
- EFS = shared, scalable, Linux-only NFS file system.
- Pay-per-use model, no pre-provisioning.
- Performance modes: General Purpose vs Max I/O.
- Throughput modes: Bursting, Provisioned, Elastic.
- Storage tiers: Standard, IA, Archive → lifecycle policies for cost optimization.
- Multi-AZ vs One Zone deployment options.

## Amazon FSx
- Launch 3rd party high-performance file systems on AWS
- Windows File Server → SMB/NTFS, AD integration, daily S3 backups.
- Lustre → HPC/ML workloads, scratch vs persistent, S3 integration.
- NetApp ONTAP → Broad compatibility, compression, de-duplication, cloning.
- OpenZFS → High performance, snapshots, cloning, no de-duplication.
- FSx = managed, scalable, specialized file systems for different workloads.

## Amazon Kinesis Data Stream
- Kinesis Data Streams = real-time ingestion & processing.
- Retention up to 365 days, replay supported.
- Partition key ensures ordering.
- Provisioned mode = manual shard scaling; On-demand mode = auto-scaling.
- Use KPL/KCL for optimized producer/consumer apps.
- Security: KMS encryption at rest, HTTPS in transit.

## Amazon data firehose
- Firehose = near real-time delivery service with buffering.
- No storage or replay → data flows directly to destinations.
- Supports AWS + 3rd-party + custom endpoints.
- Kinesis Data Streams = real-time ingestion & replay, Firehose = delivery pipeline.
- Use Lambda with Firehose for custom transformations.

## Troubleshooting kinesis data stream
- Producer issues often tied to shard limits, hot shards, batching, retries.
- Consumer issues often tied to shard scaling, iterator age, IAM permissions, Lambda limits.
- Enhanced Fan-Out is a common exam answer for read throttling.
- Always monitor iterator age, latency, CPU/memory to diagnose bottlenecks.

## imputing missing data
- Mean replacement = quick but naive.
- Median/mode = better for skewed or categorical data.
- KNN, regression, deep learning, MICE = more accurate, leverage feature relationships.
- Dropping rows = risky unless missingness is rare and unbiased.
- Best solution = collect more complete data whenever possible.
