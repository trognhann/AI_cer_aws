## EMR
- EMR = managed Hadoop + modern frameworks (Spark, Hive, Flink, etc.).
- Node types: Master (coordination), Core (tasks + HDFS), Task (tasks only).
- Transient vs long-running clusters → cost vs flexibility.
- Storage: HDFS (fast but ephemeral), EMRFS (S3-backed, persistent), EBS integration.
- Integration points: EC2, VPC, S3, CloudWatch, IAM, CloudTrail, Data Pipeline.
- Scaling: Add/remove task nodes, resize core nodes, spot instances for compute bursts.

## Missing data processing
- Mean Replacement: Replace missing data with the mean value from the rest of column (a column representation a signal feature) (fast and easy won't affect mean and sample size of overall data set)
- Medium Replacement: May be better choise of Mean Replacmentn when outliner is represent.
- Most/Mode frequents: use of category data (becase can't caculate mean/ Medium)
- Dropping row: Dropping missing row is easy but dangorous, it risk bias by removing key group.

### Imputing missing data: Machine learning
- KNN : Find the K most similar row to the one missing data
- Deep learning: powerfull with category data, require heavy computing resource and complex code, produce highly accurate result
- Regression: liner or nonliner between missing colum and other colums, use this relationship to predict and fill missing value (MICE Sota)


## Handling Unbalanced data
- Imbalance data & Accuracy Trap:
    + Race event (fraud 1/1000) are hard for model detect
    + A 'lazy' mode may always predict no Fraud.
    + This yeil 99,99% accuracy buy is unless, since it miss 100% of actual fraud case.

- Redefining Positive and Negative: Prositive mean the target event (fraud/disease), Negative mean normal/ no event

- Data lever Solution for Imbalance data
    + Oversampling: Duplicate minority class sample. Simple but effective especially for neural network.
    + undersampling: Remove majority class sample. (Not recommended)
    + SMOTE (Synthetic Minority Over-sampling Technique): Smarter approach. Uses KNN to find nearby minority samples, then averages them to create new synthetic data points.

- Threshold-level Solutions for Imbalanced Data
    + Models output probabilities, not just Yes/No. Default threshold = 50%.
    + Raise threshold (e.g., 90% → Fraud): Fewer frauds detected, more False Negatives, but fewer false alarms (False Positives).
    + Lower threshold (e.g., 30% → Fraud): Catch more frauds, fewer False Negatives, but many false alarms (False Positives).
    + Choice depends on business context and the cost of wrong predictions.


## Handling Outliers
- Variance measures squared spread from mean; standard deviation is its root, easier to interpret.
- Outlier Detection Methods
    + Standard Deviation (𝜎): In a normal distribution, most data lies within 1σ. Points beyond 2–3σ are likely outliers.
    + Boxplot (Interquartile Range, IQR): Defines outliers as points outside 1.5 × IQR from the quartiles.
